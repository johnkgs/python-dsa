# Binary Search Tree (BST)

- Buscar um elemento exige tempo O($\log n$).
- Imprimir os elementos na ordem exige apenas tempo O($\log n$).
- Encontrar o piso e o teto exige tempo O($\log n$).
- Localizar o K-ésimo menor/maior elemento exige tempo O($\log n$) quando a árvore estiver propriamente configurada.

A BST tende a funcionar melhor em situações nas quais se gasta mais tempo buscando e menos tempo construindo a árvore.

- Construir uma BST exige tempo O($n \log n$).

```python
from bintrees import BinaryTree

data = {3:'White', 2:'Red', 1:'Green', 5:'Orange', 
        4:'Yellow', 7:'Purple', 0:'Magenta'}

tree = BinaryTree(data)
tree.update({6:'Teal'})

def display_key_value(key, value):
    print('Key: ', key, 'Value: ', value)
    
tree.foreach(display_key_value)
print('Item 3 contains: ', tree.get(3))
print('The maximum item is: ', tree.max_item())
```
