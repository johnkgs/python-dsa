# Binary Heap

- Buscar um elemento exige tempo O($n$).
- Imprimir os elementos na ordem exige apenas tempo O($n \log n$).

Um heap binário tende a funcionar melhor em situações dinâmicas nas quais as chaves mudam regularmente. 

O heap binário também oferece vantagens, como descrito na lista a seguir:

- Criar as estruturas exigidas requer menos recursos porque heaps binários baseiam-se em vetores, tornando-os mais amigáveis para o cache também.

- Construir um heap binário exige tempo O($n$).

-  Usar ponteiros para implementar a árvore não é necessário.

- Contar com variações de heaps binários (por exemplo, o Heap Fibonacci) oferece vantagens, tais como o aumento e a diminuição dos tempos chave em tempo O($1$).

```python
import heapq

data = {3:'White', 2:'Red', 1:'Green', 5:'Orange', 
        4:'Yellow', 7:'Purple', 0:'Magenta'}

heap = []
for key, value in data.items():
    heapq.heappush(heap, (key, value))
heapq.heappush(heap, (6, 'Teal'))
heap.sort()

for item in heap:
    print('Key: ', item[0], 'Value: ', item[1])
print('Item 3 contains: ', heap[3][1])
print('The maximum item is: ', heapq.nlargest(1, heap))
```
