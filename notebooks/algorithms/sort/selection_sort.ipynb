# Selection Sort

- Ordenação de pior caso: O($n^2$)

A selection sort substituiu uma predecessora, a `bubble sort`.

```python
data = [9, 5, 7, 4, 2, 8, 1, 10, 6, 3]
```

```python
def selection_sort(array):
  size = len(array)
  for scan_index in range(size):
      min_index = scan_index
      
      for comparison_index in range(scan_index + 1, size):
          if array[comparison_index] < array[min_index]:
              min_index = comparison_index
              
      if min_index != scan_index:
          array[scan_index], array[min_index] = array[min_index], array[scan_index]
          print(array)

selection_sort(data)
```
