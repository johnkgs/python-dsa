# Insertion Sort

- Ordenação de melhor caso: O($n$)
- Ordenação de pior caso: O($n^2$)

```python
data = [9, 5, 7, 4, 2, 8, 1, 10, 6, 3]
```

```python
def insertion_sort(array):
  size = len(data)
  for scan_index in range(1, size):
      temp = data[scan_index]
      
      min_index = scan_index
      
      while min_index > 0 and temp < data[min_index - 1]:
          data[min_index] = data[min_index - 1]
          min_index -= 1
          
      data[min_index] = temp
      print(data)

insertion_sort(data)
```
