# Graphs

Um grafo é um tipo de extensão de uma árvore. Assim como com as árvores, há nós que se conectam uns aos outros para criar relações. Porém, diferente das árvores binárias, um grafo pode ter mais que uma ou duas conexões. Na verdade, nós grafos geralmente têm várias conexões. 

## Construindo grafos

```python
graph = {'A': ['B', 'F'],
         'B': ['A', 'C'],
         'C': ['B', 'D'],
         'D': ['C', 'E'],
         'E': ['D', 'F'],
         'F': ['E', 'A']}
```

## Percorrendo o grafo (traversing the graph)

```python
from typing import Any, Dict, List


def find_path(graph: Dict[str, List[str]], start: str, end: str, path: List[str] = []) -> List[str]:
        path = path + [start]
        
        if start == end:
            print("Ending")
            return path
        
        for node in graph[start]:
            print(f"Checking Node {node}")
            
            if node not in path:
                print(f"Path so far {path}")
                
                new_path = find_path(graph, node, end, path)
                if new_path:
                    return new_path

find_path(graph, 'B', 'E')
```
