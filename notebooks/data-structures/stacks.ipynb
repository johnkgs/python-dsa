# Stacks

Uma pilha oferece armazenamento de dados do tipo Entra por Último/Sai Primeiro `(LIFO)`.

```python
class Stack:
  def __init__(self, max_size=0):
      self.stack = []
      self.stack_size = max_size
  def display_stack(self):
    print("Stack currently contains:")
    for item in self.stack:
        print(item)
  
  def push(self, value):
    if len(self.stack) < self.stack_size:
        self.stack.append(value)
    else:
        print("Stack is full!")
  
  def pop(self):
    if len(self.stack) > 0:
        print("Popping: ", self.stack.pop())
    else:
        print("Stack is empty.")

my_stack = Stack(3)
my_stack.push(1)
my_stack.push(2)
my_stack.push(3)
my_stack.display_stack()

my_stack.push(4)

my_stack.pop()
my_stack.display_stack()

my_stack.pop()
my_stack.pop()
my_stack.pop()
```
