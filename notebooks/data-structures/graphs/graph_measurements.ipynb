```python
import networkx as nx
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
a_graph = nx.Graph()

nodes = range(1,5)
edges = [(1,2), (2,3), (3,4), (4,5), (1,3), (1,5)]

a_graph.add_nodes_from(nodes)
a_graph.add_edges_from(edges)
```

## Adicionando um novo nó e checando pelas conexões.

```python
a_graph.add_node(6)
sorted(nx.connected_components(a_graph))
```

## Adicionando uma nova aresta para o nó 6 e checando pelas conexões novamente.

```python
a_graph.add_edge(1,6)
sorted(nx.connected_components(a_graph))
```

## Determinando quais nós têm mais conexões e, portanto, são mais importantes.

```python
dict(nx.degree(a_graph)).values()
```

## Determinando quais nós têm maior probabilidade de agrupar.

```python
nx.clustering(a_graph)
```
