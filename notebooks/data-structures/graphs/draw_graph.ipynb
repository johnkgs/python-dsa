```python
import networkx as nx
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
a_graph = nx.Graph()

nodes = range(1,5)
edges = [(1,2), (2,3), (3,4), (4,5), (1,3), (1,5)]

a_graph.add_nodes_from(nodes)
a_graph.add_edges_from(edges)
```

```python
nx.draw(a_graph, with_labels=True)
```
