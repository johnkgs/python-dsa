## Grau de separação (Degrees of Separation)

```python
import networkx as nx
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
graph_data  = {'A': ['B', 'F', 'H'],
         'B': ['A', 'C'],
         'C': ['B', 'D'],
         'D': ['C', 'E'],
         'E': ['D', 'F', 'G'],
         'F': ['E', 'A'],
         'G': ['E', 'H'],
         'H': ['G', 'A']}

graph = nx.DiGraph(graph_data)
positions = nx.spring_layout(graph)
nx.draw_networkx_labels(graph, positions)
nx.draw_networkx_nodes(graph, positions)
nx.draw_networkx_edges(graph, positions)
plt.show()
```

[shortest path length](https://networkx.github.io/documentation/development/ reference/algorithms.shortest_paths.html)

```python
nx.shortest_path_length(graph, 'A')
```

```python
graph_data  = {'A': ['B', 'H'],
         'B': ['A', 'C'],
         'C': ['B', 'D'],
         'D': ['C', 'E'],
         'E': ['D', 'F', 'G'],
         'F': ['E', 'A'],
         'G': ['E', 'H'],
         'H': ['G', 'A']}

graph = nx.DiGraph(graph_data)
positions = nx.spring_layout(graph)
nx.draw_networkx_labels(graph, positions)
nx.draw_networkx_nodes(graph, positions)
nx.draw_networkx_edges(graph, positions)
plt.show()
```

```python
nx.shortest_path_length(graph, 'A')
```

## Caminhos Aleatórios (Random Walks)

```python
import random
random.seed(0)

paths = nx.all_simple_paths(graph, 'A', 'H')

path_list = []
for path in paths:
    path_list.append(path)
    print("Path Candidate: ", path)
    
sel_path = random.randint(0, len(path_list) - 1)

print("The selected path is: ", path_list[sel_path])
```
