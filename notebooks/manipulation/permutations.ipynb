## Distinguindo permutações

```python
a = np.array([1,2,3])
np.random.permutation(a)
```

```python
from itertools import permutations

a = np.array([1,2,3])

for p in permutations(a):
  print(p)
```

## Reorganizando combinações

```python
from itertools import combinations

a = np.array([1,2,3,4])

for comb in combinations(a, 2):
  print(comb)
```

```python
import random

pool = []

for comb in combinations(a, 2):
  pool.append(comb)


random.sample(pool, 3)
```

## Obtendo o Resultado Desejado Usando Recursividade

```python
def factorial(n):
  print(f"factorial called with n = {str(n)}")
  if n == 1 or n == 0:
    print("Ending condition met.")
    return 1
  else:
    return n * factorial(n-1)
    
print(factorial(5))
```

### Utilizando iteração

```python
def factorial(n):
  print(f"factorial called with n = {str(n)}")
  result = 1
  while n > 1:
    result = result * n
    n = n - 1
    print(f"Current value of n is {str(n)}")
  print("Ending condition met.")
  return result
print(factorial(5))
```
