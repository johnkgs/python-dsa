## Multiplicando matrizes

```python
a = np.array([[1,2,3],[4,5,6]])
b = np.array([[1,2,3],[4,5,6]])
a * b
```

## Produto Escalar

```python
a = np.array([[1,2,3],[4,5,6]])
b = np.array([[1,2,3],[3,4,5],[5,6,7]])
a.dot(b)
```

```python
a = np.matrix([[1,2,3],[4,5,6]])
b = np.matrix([[1,2,3],[3,4,5],[5,6,7]])
a*b
```

## Definindo operações matriciais avançadas

```python
change_it = np.array([1,2,3,4,5,6,7,8]) 
change_it
```

```python
change_it.reshape(2,4)
```

```python
change_it.reshape(2,2,2) 
```

```python
change_it = change_it.reshape(2,4)
np.transpose(change_it) 
```

```python
np.identity(4)
```

```python
a = np.array([[1,2], [3,4]])
b = np.linalg.inv(a)
np.allclose(np.dot(a,b), np.identity(2))
```
