```python
import pandas as pd
import numpy as np
```

## Removendo linhas duplicadas

```python
df = pd.DataFrame({"A": [0,0,0,0,0,1,0], "B": [0,2,3,5,0,2,0], "C": [0,3,4,1,0,2,0]})

print(df, "\n")

df = df.drop_duplicates()

print(df)
```

## Lidando com valores ausente

```python
df = pd.DataFrame({"A": [0,0,1,None], "B": [1,2,3,4], "C": [np.NAN,3,4,1]}, dtype=int)
print(df, "\n")
values = pd.Series(df.mean(), dtype=int)

print(values, "\n")

df = df.fillna(values)
print(df)
```
