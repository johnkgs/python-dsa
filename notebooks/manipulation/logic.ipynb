```python
import numpy as np
```

## Operações

```python
a = np.array([1, 2, 3, 4])
b = np.array([2, 2, 4, 4])
```

```python
a == b
```

```python
a < b
```

```python
a = np.array([True, False, True, False])
b = np.array([True, True, False, False])
```

```python
np.logical_or(a, b)
```

```python
np.logical_and(a, b)
```

```python
np.logical_not(a)
```

```python
np.logical_xor(a, b)
```
