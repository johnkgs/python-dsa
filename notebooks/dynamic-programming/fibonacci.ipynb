## Fibonacci numbers


```python
def fib(n, tab=0):
    if n == 0:
        return 0
    elif n == 1:
        return 1
    else:
        print("lvl %i, Summing fib(%i) and fib(%i)" % (tab, n - 1, n - 2))
        return fib(n - 1, tab + 1) + fib(n - 2, tab + 1)
```

```python
fib(7)
```

```python
memo = dict()


def fib_mem(n, tab=0):
    if n == 0:
        return 0
    elif n == 1:
        return 1
    else:
        if (n - 1, n - 2) not in memo:
            print("lvl %i, Summing fib(%i) and fib(%i)" % (tab, n - 1, n - 2))
            memo[(n - 1, n - 2)] = fib_mem(n - 1, tab + 1) + fib_mem(n - 2, tab + 1)
        return memo[(n - 1, n - 2)]
```

```python
fib_mem(7)
```

## Memoization


```python
def fib(n):
    if n == 0:
        return 0
    elif n == 1:
        return 1
    else:
        return fib(n - 1) + fib(n - 2)
```

```python
%timeit -n 1 -r 1 print(fib(36))
```

```python
from functools import lru_cache


@lru_cache(maxsize=None)
def fib(n):
    if n == 0:
        return 0
    elif n == 1:
        return 1
    else:
        return fib(n - 1) + fib(n - 2)
```

```python
%timeit -n 1 -r 1 print(fib(36))
```
