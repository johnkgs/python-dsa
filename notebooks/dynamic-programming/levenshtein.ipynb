## Levenshtein Distance


```python
import numpy as np
import pandas as pd
```

```python
s1 = "Saturday"
s2 = "Sunday"
m = len(s1)
n = len(s2)
D = np.zeros((m + 1, n + 1))
D[0, :] = list(range(n + 1))
D[:, 0] = list(range(m + 1))
```

- **Exclusão**: Remover uma letra de uma palavra.
- **Inserção**: Inserir uma letra em uma palavra e obter outra palavra.
- **Substituição**: Substituir uam letra por outra, como trocar a letra p por c para obter cão de pão.

```python
for j in range(1, n + 1):
    for i in range(1, m + 1):
        if s1[i - 1] == s2[j - 1]:
            D[i, j] = D[i - 1, j - 1]
        else:
            D[i, j] = np.min(
                [
                    D[i - 1, j] + 1,  # a deletion
                    D[i, j - 1] + 1,  # an insertion
                    D[i - 1, j - 1] + 1,  # a substitution
                ]
            )
print("Levenshtein distance is %i" % D[-1, -1])
```

```python
pd.DataFrame(D, index=list(" " + s1), columns=list(" " + s2))
```
